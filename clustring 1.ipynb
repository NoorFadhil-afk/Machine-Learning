```python
marks = [35, 40, 42, 60, 65, 70]
k = 2

# Initial centroids
c1 = marks[0]
c2 = marks[1]

for _ in range(5):
    cluster1 = []
    cluster2 = []

# Assign marks to nearest centroid
    for m in marks:
        if abs(m - c1) < abs(m - c2):
            cluster1.append(m)
        else:
            cluster2.append(m)

# Update centroids
    c1 = sum(cluster1) / len(cluster1)
    c2 = sum(cluster2) / len(cluster2)

print("Cluster 1:", cluster1)
print("Cluster 2:", cluster2)
print("Centroids:", c1, c2)
```
